## Quick project info form

Fill out the form below so the agent can capture your name, contact, and project goal before proceeding. Run the cell to render the form, enter your details, and click **Submit** to record them right in the notebook output.


```python
import ipywidgets as widgets
from IPython.display import display, clear_output

name_input = widgets.Text(description='Name', placeholder='Your name', style={'description_width': 'initial'})
email_input = widgets.Text(description='Email', placeholder='you@example.com', style={'description_width': 'initial'})
goal_input = widgets.Textarea(
    description='Project goal',
    placeholder='What do you want the agent to help with?',
    layout=widgets.Layout(width='500px', height='120px'),
    style={'description_width': 'initial'},
)
submit_btn = widgets.Button(description='Submit', button_style='success', tooltip='Submit your project info')
output = widgets.Output()

def handle_submit(_):
    with output:
        clear_output()
        print('Thanks for the details! Here is what you submitted:')
        print(f"  Name: {name_input.value.strip() or '(not provided)'}")
        print(f"  Email: {email_input.value.strip() or '(not provided)'}")
        print('  Project goal:')
        if goal_input.value.strip():
            print(f"    {goal_input.value.strip()}")
        else:
            print('    (not provided)')

submit_btn.on_click(handle_submit)
form = widgets.VBox([name_input, email_input, goal_input, submit_btn])
display(form, output)
```
